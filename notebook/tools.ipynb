```python
from groq import BadRequestError
import time

def invoke_with_retry(agent, payload, max_retries=2):
    for attempt in range(max_retries + 1):
        try:
            return agent.invoke(payload)
        except BadRequestError as e:
            if "tool_use_failed" in str(e) and attempt < max_retries:
                time.sleep(0.5)
                continue
            raise
```

```python
messages = [{"role": "user", "content": "suggest me a python job"}]
result1 = invoke_with_retry(agent, {"messages": messages})
result1["messages"][-1].content
```

```python
# reuse result1's message history so job_ids from search_jobs are still in context
messages = result1["messages"]
messages.append({"role": "user", "content": "tell me more about Senior Python Developer role"})
result2 = invoke_with_retry(agent, {"messages": messages})
result2["messages"][-1].content
```
